```python
import os
import matplotlib.pyplot as plt
import json
from constants import REPO_ROOT

os.chdir(REPO_ROOT)
```

```python
ARTIFACT_NAME = "20261101T143124Z_c8abdba6"
ARTIFACT_PATH = (
    f"artifacts/delta/train/uniform_excluding_evaluation_regions/{ARTIFACT_NAME}"
)
UPDATES_PATH = f"{ARTIFACT_PATH}/updates.jsonl"
EPISODES_PATH = f"{ARTIFACT_PATH}/episodes.jsonl"

ACTOR_LOSS = "actor_loss"
CRITIC_LOSS = "critic_loss"
EPISODE_REWARD = "episode_reward"
```

```python
updates = []
with open(UPDATES_PATH, "r") as f:
    for line in f:
        updates.append(json.loads(line))

episodes = []
with open(EPISODES_PATH, "r") as f:
    for line in f:
        episodes.append(json.loads(line))
```

```python
# Plot actor and critic loss over time
actor_losses = [update[ACTOR_LOSS] for update in updates]
critic_losses = [update[CRITIC_LOSS] for update in updates]

plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(actor_losses, label="Actor Loss", color="blue")
plt.title("Actor Loss Over Time")
plt.xlabel("Update Step")
plt.ylabel("Loss")
plt.subplot(1, 2, 2)
plt.plot(critic_losses, label="Critic Loss", color="orange")
plt.title("Critic Loss Over Time")
plt.xlabel("Update Step")
plt.ylabel("Loss")
plt.tight_layout()
plt.show()
```

```python
# Plot episode rewards as training continues
episode_rewards = [episode[EPISODE_REWARD] for episode in episodes]

plt.figure(figsize=(12, 6))
plt.plot(episode_rewards, label="Rewards", color="blue")
plt.title("Rewards Over Time")
plt.xlabel("Update Step")
plt.ylabel("Reward")
plt.tight_layout()
plt.show()
```
